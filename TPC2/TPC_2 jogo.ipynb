### TPC2: Adivinha o número
- Crie um programa em Python para jogar o jogo "Adivinha o número":
    - O jogo pode ter 2 modalidades: computador pensa num número (entre 0 e 100), utilizador tenta adivinhar; ou, o utilizador pensa num número (entre 0 e 100) e o computador tenta adivinhar;
    - Quem tenta adivinhar responde com uma das afirmações: "Acertou", "O número que pensei é Maior" ou "O número que pensei é Menor";
    - Uma vez descoberto o número o programa deverá terminar imprimindo o número de tentativas que quem adivinhou usou para chegar ao resultado.

```python
import random

def adivinha_numero():
    tentativas=0
    escolha = input("Queres que seja eu a escolher o número ou preferes ser tu?: (computador/utilizador)")

    if escolha == "computador":
        n = random.randint(1, 100)
        op1=int(input("Já escolhi um número que número achas que escolhi?: "))
        while op1 != n:
            if op1>n:
                print(f"O número que pensei é menor que {op1}")
                tentativas+=1
                print(f"O número de tentativas é {tentativas}")
                op1=int(input("Tenta outra vez: ")) 
            elif op1<n:
                print(f"O número que pensei é maior que {op1}")
                tentativas+=1
                print(f"O número de tentativas é {tentativas}")
                op1=int(input("Tenta outra vez: ")) 
        print(f"Acertaste o numero que escolhi foi {n} com {tentativas} tentativas")

    elif escolha == "utilizador":
        op2=int(input("Escolhe um número que eu tento adivinhar: "))
        n = 50
        teto=101
        chao=0
        while n != op2:
            if n>op2:
                print(f"O número é menor que {n}")
                tentativas+=1
                print(f"O número de tentativas é {tentativas}")
                teto=n
                n=int((teto+chao)/2)
            elif n<op2:
                print(f"O número é maior que {n}")
                tentativas+=1
                print(f"O número de tentativas é {tentativas}")
                chao=n
                n=int((teto+chao)/2)
        print(f"Acertei o numero que escolheste foi {op2} com {tentativas} tentativas")
    else:
        print("A tua escolha tem de ser válida (computador/utilizador)")
        

adivinha_numero()
```

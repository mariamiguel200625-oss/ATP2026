### TPC3: Corrida para o 100

O total começa em 0. O jogador e o computador alternam somando um número de 1 a 10 ao total. Quem atingir exatamente o número 100 vence.
- Se o computador jogar primeiro deverá vencer;
- Se o computador jogar em segundo poderá vencer se o outro jogador fizer uma aposto fora da estratégia vencedora.

Implementa o jogo com 2 vertentes: o computador joga primeiro (deverá ganhar sempre), e o computador joga em segundo lugar (poderá ganhar ou não dependendo das jogadas do outro).

```python
import random

jogador = input("Queres jogar em primeiro ou em segundo lugar? ").lower()
total = 0

if jogador == "segundo":
    total = 1
    print("O computador somou 1. Total: 1")

if jogador == "primeiro" or jogador == "segundo":
    while total < 100:
        # Jogada do jogador
        jogada = int(input(f"Total: {total}. Quanto queres somar (1-10)? "))
        while jogada < 1 or jogada > 10 or total + jogada > 100:
            jogada = int(input("Jogada inválida. Tenta outra vez (1-10): "))
        total += jogada
        print(f"Somaste {jogada}. Total: {total}")

        if total == 100:
            print("Ganhaste!")
        else:
            # Jogada do computador
            jogada_pc = (1 - total) % 11
            if jogada_pc == 0:
                jogada_pc = random.randint(1, 10)
            total += jogada_pc
            print(f"O computador somou {jogada_pc}. Total: {total}")

            if total == 100:
                print("O computador ganhou!")
else:
    print("Escolhe direito.")
```
